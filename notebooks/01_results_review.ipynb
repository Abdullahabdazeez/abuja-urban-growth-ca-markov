# Abuja CA–Markov Results Review

This notebook reviews the final 2025–2035 tables supplied with the project.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
TABLES = ROOT / "data/processed/tables"
```

```python
area = pd.read_csv(TABLES / "Abuja_LULC_Area_2025_2035.csv")
area
```

```python
pivot = area.pivot(index="class_name", columns="year", values="area_sqkm")
pivot["net_change_sqkm"] = pivot[2035] - pivot[2025]
pivot
```

```python
pivot[[2025, 2035]].plot(kind="bar", figsize=(10, 5))
plt.ylabel("Area (km²)")
plt.title("Abuja LULC area, 2025 and 2035")
plt.tight_layout()
```
